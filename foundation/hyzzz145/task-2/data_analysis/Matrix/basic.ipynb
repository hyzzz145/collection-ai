```python
import numpy as np
```

创建一个形状为 (10, 10) 的 NumPy 数组，命名为 data_matrix，使其包含从 0 到 99 的连续整数。

```python
data_matrix = np.arange(0,100)
data_matrix = data_matrix.reshape(10,-1)
print(data_matrix)
```

子矩阵提取：从 data_matrix 中提取出位于中心的 4x4 子矩阵。

```python

data__matrix_center = data_matrix[4:6,4:6]
print(data__matrix_center)
```

条件筛选与赋值

```python
data_matrix[data_matrix>75] = 0
print(data_matrix)
```

向量化运算

```python
data_matrix = 0.8*data_matrix
print(data_matrix)
```

数据聚合与定位：

- 找出最终矩阵中的最大值。
- 确定该最大值在矩阵中的行和列索引。
- 提示：可结合使用 numpy.argmax 和 numpy.unravel_index 来高效完成定位

```python
max = np.max(data_matrix)
print(max)
idx = np.argmax(data_matrix)
row,column = np.unravel_index(idx,data_matrix.shape)
print(f'row={row},column={column}')
```
